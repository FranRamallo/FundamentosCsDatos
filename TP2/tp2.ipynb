```python
# Cuantos directores distintos hay
#Primero separo las peliculas que tienen mas de un director
directores = leer_datos_netflix["director"].str.split(",").explode().str.strip()
cant_directores_distintos = directores.nunique()
print(cant_directores_distintos)
```

```python
# Quien es el director con mas peliculas
# Cuento cuantas veces aparece el mismo director
director_cant_peliculas = directores.value_counts()
# Busco el maximo (el que mas salio)
director_mas_peliculas = director_cant_peliculas.idxmax()
cant_peliculas_del_director = director_cant_peliculas.max()
print(f"El director con mas peliculas es {director_mas_peliculas} con {cant_peliculas_del_director} peliculas")
```

```python
# Cuantas series series hay y cuantos programas de TV
cant_tv_shows_y_movies = leer_datos_netflix["type"].value_counts()
print(cant_tv_shows_y_movies)
```

```python
# Pelicula y serie mas vieja del catalogo
peliculas = leer_datos_netflix[leer_datos_netflix["type"] == "Movie"]
pelicula_mas_vieja = peliculas.loc[peliculas["release_year"].idxmin()]
series = leer_datos_netflix[leer_datos_netflix["type"] == "TV Show"]
serie_mas_vieja = series.loc[series["release_year"].idxmin()]
print(peliculas)
print(pelicula_mas_vieja)
print()
print(serie_mas_vieja)
```

```python
# Dado un país, determinar cuántas películas de ese país hay en la base de datos.
pais = input("Ingrese el nombre de un pais: ")
peliculas_del_pais = leer_datos_netflix[(leer_datos_netflix["type"] == "Movie") & (leer_datos_netflix["country"].str.contains(pais, na=False))]

print(f"Hay {len(peliculas_del_pais)} peliculas de {pais}")
```

```python
# Cuanto contenido tiene cada plataforma
contenido_por_plataforma = leer_streaming_arg["platform"].value_counts()
print(contenido_por_plataforma)

# Y cual es la plataforma con mas contenido
plataforma_mas_contenido = contenido_por_plataforma.idxmax()
cant_plataforma_mas_contenido = contenido_por_plataforma.max()
print(f"La plataforma con mas contenido es {plataforma_mas_contenido} con {cant_plataforma_mas_contenido}")
```

```python
# Cuantas peliculas y series hay
cant_peliculas_series = leer_streaming_arg["type"].value_counts()
print(cant_peliculas_series)
```

```python
plataforma = input("Ingrese una plataforma: ")

if plataforma in contenido_por_plataforma.index:
    contenido_plataforma = contenido_por_plataforma[plataforma]
    print(f"{plataforma} tiene {contenido_plataforma} contenidos.")
else:
    print(f"No se encontró la plataforma {plataforma}.")
```

```python
# Plataforma con mayor catidad de peliculas
peliculas_streaming = leer_streaming_arg[leer_streaming_arg["type"] == "MOVIE"]
peliculas_por_plataforma = peliculas_streaming.groupby("platform").size()
platform_max = peliculas_por_plataforma.idxmax()
cantidad_max = peliculas_por_plataforma.max()
print(f"La plataforma con mas peliculas es {platform_max} con {cantidad_max} peliculas")
```

```python
# Plataforma con mayor cantidad de series
series_streaming = leer_streaming_arg[leer_streaming_arg["type"] == "SHOW"]
series_por_plataforma = series_streaming.groupby("platform").size()
platform_max_series = series_por_plataforma.idxmax()
cantidad_max_series = series_por_plataforma.max()
print(f"La plataforma con mas peliculas es {platform_max_series} con {cantidad_max_series} series")
```

```python
# Plataforma que tiene, en promedio, contenidos más popular
promedio_por_plataforma = leer_streaming_arg.groupby("platform")["tmdbPopularity"].mean()
print(promedio_por_plataforma)
```
